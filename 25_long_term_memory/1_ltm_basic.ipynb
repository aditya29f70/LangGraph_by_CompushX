```python
from langgraph.store.memory import InMemoryStore
```

```python
# creating a store
store= InMemoryStore()
```

```python
# creating a namespace
namespace= ('user', 'u1')
```

## Creating Memories

```python
store.put(namespace, "1", {"data":"User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})
```

```python
# another namespace
namespace2= ("user", 'u2')


## add memory inside that namespace

store.put(namespace2, "1", {"data":"User likes pasta"})
store.put(namespace2, "2", {"data":"User prefers grid style navigation"})
```

## Retrieving Memories (fetch a particular memory using its key)

```python
# store.get(namespace, key)

store.get(namespace, "1")
```

## Retrieving all Memories

```python
items= store.search(namespace2)

for item in items:
    print(item.value)
```

## Semantic Search

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding= HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
store= InMemoryStore(index={"embed":embedding, "dim":384}) # now it is specialised memory store ; which has semantic search capability
```

```python
# creating namespace
namespace= ("users", 'u1')
```

```python
# creating memory inside that namespace
store.put(namespace, "1", {"data":"User prefer concise answers over long explanations"})
store.put(namespace, "2", {"data":"User like examples in python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data":"User prefer dark mode in application"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislike overly theoretical explanation"})
store.put(namespace, "7", {"data": "User is based in India"})
store.put(namespace, "8", {"data": "User is prefer step-by-step resoning"})
store.put(namespace, "9", {"data":"User likes real world analogies"})
store.put(namespace, "10", {"data":"User prefers bullet points over paragraphs"})

# these will be stored with its text as well as its embedding so now we will use 'search' method with query so store will convert query to embedding will do semantic search and retrieve given numbers of similar memories
```

```python
items= store.search(namespace, query="what is user currently learning?", limit=3)


for item in items:
    print(item.value)
```

```python
items= store.search(namespace, query="what are user preferences", limit=3)

for item in items:
    print(item.value)
```
